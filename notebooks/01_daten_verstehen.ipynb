```python
def features_fuer_song(song):
    erster_play = song["endTime"].min()
    ende_woche = erster_play + pd.Timedelta(days=7)
    ende_monat = erster_play + pd.Timedelta(days=30)

    erste_woche = song[(song["endTime"] < ende_woche) & (song["gehoert"] == True)]
    erster_monat = song[(song["endTime"] < ende_monat) & (song["gehoert"] == True)]

    return {
        "first_played_at": erster_play,
        "plays_first_7d": len(erste_woche),
        "is_favorite": len(erster_monat) >= 10,
        "woche_fertig": ende_woche <= ende_daten,
        "label_fertig": ende_monat <= ende_daten,
    }
```

```python
features_fuer_song(song)
```

```python
anderer_song = df[df["song_id"] == "Naka 99 33 - Mond herz 837"]
features_fuer_song(anderer_song)
```

```python
zeilen = []

for song_id, song in df.groupby("song_id"):
    resultat = features_fuer_song(song)
    resultat["song_id"] = song_id
    zeilen.append(resultat)

features = pd.DataFrame(zeilen)
features.head()
```
